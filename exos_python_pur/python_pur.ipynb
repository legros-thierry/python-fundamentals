```python
phrases = [
    "Python c'est cool",
    "Je progresse tous les jours",
    "IA et data",
    "VS Code dans le navigateur",
    "Je vais faire un bootcamp"
]

listlongueur = [len(p) for p in phrases]
print(listlongueur)
```

```python
phrases_longues = [p for p in phrases if len(p) >15]
print(phrases_longues)
```

```python
# Maintenant, construis un dictionnaire phrases_par_longueur qui associe chaque phrase à sa longueur,
# mais seulement pour les phrases de plus de 10 caractères.

# Structure mentale :
# “pour chaque p dans phrases, associe p à len(p) si len(p) > 10”.

phrases_par_longueur = {p:len(p) for p in phrases if len(p) > 10}
print(phrases_par_longueur)
```

```python
assert "IA et data" not in phrases_par_longueur, "assert 1 casse !"
assert phrases_par_longueur["Je vais faire un bootcamp"] == 25, "assert 2 casse !!"
```

```python
def stats_liste(nums):
  return({"min":min(nums), "max":max(nums), "moyenne":sum(nums)/len(nums)})

dico = stats_liste([1,2,3,10])
print(dico)
assert int(dico["moyenne"]) == 4, "assert casse !"
assert stats_liste([5, 5, 5])["moyenne"] == 5
assert stats_liste([-1, 1])["min"] == -1
assert stats_liste([-1, 1])["max"] == 1



```

```python
tickets = [
    {"id": "T1", "statut": "ouvert", "priorite": 3},
    {"id": "T2", "statut": "clos",   "priorite": 5},
    {"id": "T3", "statut": "ouvert", "priorite": 4},
    {"id": "T4", "statut": "en cours", "priorite": 2},
]

def ids_ouverts(tickets):
  return([t["id"] for t in tickets if t["statut"]=="ouvert"])

print(ids_ouverts(tickets))
```

```python
def tickets_haute_priorite(tickets, seuil):
  return([t for t in tickets if t["priorite"]>=seuil])

res = tickets_haute_priorite(tickets, 4)
print(res)
```

```python
lignes = [
    "2024-01-01 INFO Démarrage application",
    "2024-01-01 ERROR Base de données indisponible",
    "2024-01-01 INFO Requête utilisateur",
    "2024-01-02 WARNING Espace disque faible",
    "2024-01-02 ERROR Timeout API externe",
]

def filtrer_lignes_par_mot_cle(lignes, mot_cle):
    return([lmotcle for lmotcle in lignes if mot_cle in lmotcle])

print(filtrer_lignes_par_mot_cle(lignes, "ERROR"))
```

```python
lignes = [
    "2024-01-01 INFO Démarrage application\n",
    "2024-01-01 ERROR Base de données indisponible\n",
    "2024-01-01 INFO Requête utilisateur\n",
    "2024-01-02 WARNING Espace disque faible\n",
    "2024-01-02 ERROR Timeout API externe\n",
]

with open("logs_bruts.txt", "w", encoding="utf-8") as f:
    f.writelines(lignes)
```

```python
def filtrer_fichier_vers_liste(chemin, mot_cle):

  with open(chemin, "r", encoding="utf-8") as f:
    return([lmotcle for lmotcle in f if mot_cle in lmotcle])

erreurs = filtrer_fichier_vers_liste("logs_bruts.txt", "ERROR")
print(erreurs)
```

```python
def filtrer_fichier_vers_fichier(chemin_entree, chemin_sortie, mot_cle):
    with open(chemin_entree, "r", encoding="utf-8") as e:
      with open(chemin_sortie, "w", encoding="utf-8") as s:
        lfiltrees = filtrer_lignes_par_mot_cle(e,mot_cle)
        s.writelines(lfiltrees)
    return len(lfiltrees)

nb = filtrer_fichier_vers_fichier("logs_bruts.txt", "logs_errors.txt", "ERROR")
print(nb)
```
